### Coin Change Problem (Minimum Coins)

This problem aims to find the minimum number of coins from a given set of denominations that sum up to a target amount. We use dynamic programming to build a table of solutions for all amounts from 0 to the target.

```python
def coin_change(coins, amount):
    # dp[i] will be storing the minimum number of coins required for amount i
    # Initialize with a value larger than any possible number of coins (amount + 1)
    dp = [float('inf')] * (amount + 1)
    dp[0] = 0

    # coin_used stores which coin was last used to reach this amount (for reconstruction)
    coin_used = [-1] * (amount + 1)

    for i in range(1, amount + 1):
        for coin in coins:
            if i - coin >= 0:
                if dp[i - coin] + 1 < dp[i]:
                    dp[i] = dp[i - coin] + 1
                    coin_used[i] = coin

    if dp[amount] == float('inf'):
        return None, []

    # Reconstruct the coins used
    result_coins = []
    temp_amount = amount
    while temp_amount > 0:
        result_coins.append(coin_used[temp_amount])
        temp_amount -= coin_used[temp_amount]

    return dp[amount], result_coins

# Example usage:
coin_denominations = [1, 5, 10, 25]
target_amount = 63

min_coins, used = coin_change(coin_denominations, target_amount)

if min_coins:
    print(f"Target Amount: {target_amount}")
    print(f"Minimum coins needed: {min_coins}")
    print(f"Coins used: {used}")
else:
    print("No solution possible with the given denominations.")
```
